````python
from posydon.popsyn.synthetic_population import Population

files = {
    "1 Z☉": "../Generated_Tuto_Data/1e+00_Zsun_population.h5",
    "0.1 Z☉": "../Generated_Tuto_Data/1e-01_Zsun_population.h5",
    "0.01 Z☉": "../Generated_Tuto_Data/1e-02_Zsun_population.h5",
    "0.001 Z☉": "../Generated_Tuto_Data/1e-03_Zsun_population.h5",
    "2 Z☉": "../Generated_Tuto_Data/2e+00_Zsun_population.h5",
    "0.2 Z☉": "../Generated_Tuto_Data/2e-01_Zsun_population.h5",
}

for label, filename in files.items():
    pop = Population(filename)
    print(f"\n Population containing binaries with {label}:")
    print(pop.mass_per_metallicity)
    print("=" * 50)
````
