```python
from posydon.popsyn.synthetic_population import Population
pop = Population('../Generated_Tuto_Data/1e+00_Zsun_population.h5')
```

```python
pop.mass_per_metallicity
```

1. The index (metallicity) is the metallicity of your population in solar units.
2. <u>simulated mass</u>: The total ZAMS mass that has been evolved in the population.
3. <u>simulated_mass_single</u>: The total ZAMS mass from initially single stars.
4. <u>simulated_mass_binaries</u>: The total ZAMS mass from initially binary stars.
5. <u>number_of_systems</u> Shows the number of systems in the file.

```python
# you can also access the total number of systems in the file with
print("The number of system is :", pop.number_of_systems)
```

### 2. Select a sub-population

The Population.export_selection() function will export just the indices of the binaries you’re interested in into a new file. The simulated mass will remain the same since they are dependent on the population run.

```python
indices = [0, 9]

# Creates a new file called "selected.h5" with only the indices selected
pop.export_selection(indices, '../Generated_Tuto_Data/selection.h5', overwrite=True)
```

```python
selected = Population('..//Generated_Tuto_Data/selection.h5')
selected.mass_per_metallicity
```

```python
print('Selected simulated mass : ', np.sum(selected.oneline[['S1_mass_i', 'S2_mass_i']].to_numpy()))
```

- To create multi-metallicity populations file, we can do : 

```python
pop.export_selection(indices, '../Generated_Tuto_Data/selected.h5', overwrite=True)

selected = Population('..//Generated_Tuto_Data/selected.h5')
selected.mass_per_metallicity
```

```python
# This will add to the file and add the extra simulated mass
pop.export_selection(indices, '../Generated_Tuto_Data/selected.h5', append=True)
selected = Population('../Generated_Tuto_Data/selected.h5')
selected.mass_per_metallicity
```

-> In this last example, there is twice each systems (2 systems at first then double it).
